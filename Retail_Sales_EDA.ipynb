```python
df['Date']=pd.to_datetime(df['Date'], format='%Y-%m-%d')
```

```python
print(df['Date'].dtype)
```

```python
df['Month'] = df['Date'].dt.month # Extract month drom date
```

```python
monthly_sales = df.groupby('Month')['Total Amount'].sum()
```

```python
import matplotlib.pyplot as plt 

plt.plot (monthly_sales.index,monthly_sales.values)
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales Trend")
plt.show()
```

**Observation:** Monthly sales fluctuate across the months, showing variation in sales performance over time.

```python
df['Quarter'] =df['Date'].dt.to_period('Q').astype(str)
```

```python
quarterly_sales = df.groupby('Quarter')['Total Amount'].sum()
```

```python
quarterly_sales
```

```python
import matplotlib.pyplot as plt
plt.plot (quarterly_sales.index,quarterly_sales.values,marker='o')
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.title("Quarterly Sales Treand")
plt.show()
```

**Observation:** Q2 and Q4 of 2023 recorded relatively higher sales compared with Q1 and Q3.

```python
gender_count = df['Gender'].value_counts()
```

```python
gender_count
```

```python
gender_count.plot(kind='bar')

plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.title("Gender Distribution")
plt.show()
```

**Observation:** Female and male customers are almost evenly distributed.

```python
df['Age Group'] = pd.cut(df['Age'],bins=[17,25,35,50,64],labels=['18-25', '26-35', '36-50', '51-64'])
```

```python
age_group_count = df['Age Group'].value_counts().sort_index() 
```

```python
age_group_count
```

```python
age_group_count.plot (kind='bar')

plt.xlabel ("Age Group")
plt.ylabel ("Number of customer")
plt.title("Age Group Distribution")
plt.show()
```

**Observation:** The 36–50 and 51–64 age groups have the highest customer counts.

```python
category_quantity = df.groupby('Product Category')['Quantity'].sum().sort_values( ascending = False)

                                                                                
```

```python
category_quantity
```

```python
category_quantity.plot(kind='bar')

plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")
plt.title("Sales Quantity by Product Category")
plt.show()
```

**Observation:** The quantity sold varies across different product categories.

```python
category_revenue = df.groupby('Product Category')['Total Amount'].sum().sort_values(ascending=False)
```

```python
average_spending_gender = df.groupby('Gender')['Total Amount'].mean()
```

```python
average_spending_gender.plot(kind='bar')

plt.xlabel("Gender")
plt.ylabel("Average Spending")
plt.title("Average Spending by Gender")
plt.show()
```

**Observation:** Average spending is similar for male and female customers.
