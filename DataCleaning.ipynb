## 1. Data Quality Report

### Data Quality Findings

The dataset contains 891 records. Numerical columns were checked for their statistical ranges and potential anomalies. Age ranges from 0.42 to 80 years, while Fare ranges from 0 to 512.3292. These ranges were further examined using the IQR method for outlier detection.

## 2. Missing Data Handling

### Age

Missing Age values were replaced with the median because Age is a numerical variable and the median is less affected by extreme values.

### Embarked

Missing Embarked values were replaced with the mode because Embarked is a categorical variable, and the mode represents the most frequently occurring category.

### Cabin

The Cabin column contains a large number of missing values. Instead of removing these rows, the missing values were replaced with "Unknown" to preserve the available records.

## 3.Duplicate Removal

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:",duplicate_count)
```

No duplicate rows were found in the dataset,so no duplicate records were removed.

## 4. Standardisation

```python
print("Sex valiues:")
print(df["Sex"].unique())

print("\nEmbarked values:")
print(df["Embarked"].unique())
```

The categorical values were checked for consistency. The values in the Sex and Embarked columns were already in a consistent format, so no standardisation was required.

## 5. Outlier Detection

### Age Outliers

```python
Q1_age = df["Age"].quantile(0.25)
Q3_age = df["Age"].quantile(0.75)

IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

age_outliers = df[(df["Age"] < lower_age) | (df["Age"] > upper_age)]

print("Lower limit:", lower_age)
print("Upper limit:", upper_age)
print("Number of Age outliers:", age_outliers.shape[0])
```

66 Age values were identified as statistical outliers using the IQR method. These values were retained because the ages are valid observations and are not necessarily data errors.

### Fare Outliers

```python
Q1_fare = df["Fare"].quantile(0.25)
Q3_fare = df["Fare"].quantile(0.75)

IQR_fare = Q3_fare - Q1_fare

lower_fare = Q1_fare - 1.5 * IQR_fare
upper_fare = Q3_fare + 1.5 * IQR_fare

fare_outliers = df[(df["Fare"] < lower_fare) | (df["Fare"] > upper_fare)]

print("Lower limit:", lower_fare)
print("Upper limit:", upper_fare)
print("Number of Fare outliers:", fare_outliers.shape[0])
```

116 Fare values were identified as statistical outliers using the IQR method. These values were retained because high fare values can represent valid ticket prices and are not necessarily data errors.
